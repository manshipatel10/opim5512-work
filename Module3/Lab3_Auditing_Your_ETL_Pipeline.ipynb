## 3b. Measure the extractor's accuracy

````python
FIELDS = ["price", "year", "make", "model", "mileage"]


def same(extracted_value, true_value, field):
    if field in ["make", "model"]:
        return str(extracted_value).strip().lower() == str(true_value).strip().lower()
    try:
        return float(extracted_value) == float(true_value)
    except (TypeError, ValueError):
        return False


def score(table, truth):
    rows = []
    for field in FIELDS:
        right = 0
        for pid in truth:
            table_row = table[table["post_id"] == pid].iloc[0]
            if same(table_row[field], truth[pid][field], field):
                right = right + 1
        rows.append({"field": field, "correct": right, "out_of": len(truth), "accuracy": right / max(len(truth), 1)})
    return pd.DataFrame(rows)


before = score(cars, truth)
print(before.to_string())
````

# Part 4 · Fix it: three ways to get make and model (30 min)

The clue: a line that is **only a 4-digit year**, and the **next line** starts with the **make** then the **model**.
Three ways to use that clue, from easiest to fanciest.

## Way 1 · Plain Python, no regex (given: read it, run it)
This is the 3c anchor turned into a function. Walk the lines. When a line is exactly 4 digits and starts with 19 or 20,
split the next line into words: the first word is the make, the second is the model.
The test prints all 5 files next to what the extractor said. **Watch files 03 and 04**: the extractor said
`Contact Information` for both. Does Way 1 get them right?

````python
def way1_make_model(text):
    ad_lines = text.split("\n")
    for i in range(len(ad_lines) - 1):
        line = ad_lines[i].strip()
        if len(line) == 4 and line.isdigit() and (line.startswith("19") or line.startswith("20")):
            words = ad_lines[i + 1].strip().lower().split()
            if len(words) >= 2:
                return words[0], words[1]
    return "", ""


for i in range(len(graded)):
    text = read_ad_text(graded.iloc[i]["source_txt"])
    print("file", str(i + 1).zfill(2), "| extractor said:", graded.iloc[i]["make"], graded.iloc[i]["model"], "| way 1:", way1_make_model(text))
````

**Why isn't Way 1 perfect?** The code does exactly what it is told; the data is messy. Only the **year** comes from
Craigslist's form. The make/model line is **typed by the seller**, so it carries their typos (`Hyunda`, `FOR FOCUS`),
extra words (`Honda Accord LX cylinders`), missing words (`Pontiac` alone), and **multi-word names** that "first word,
second word" cannot handle:
- two-word **models**: file 09 is `bmw 3 series 320i xdrive`, so Way 1 says the model is `3` (others: `grand cherokee`, `town & country`, `super duty`)
- two-word **makes**: `land rover`, `alfa romeo`, `aston martin`, `rolls royce` (none in Hartford, but maybe in your area)

Our truth-table rule (model = first word) keeps grading simple, but it is a simplification. Way 3 tackles typos; a
list of multi-word names (check them first) or an LLM (A07) can tackle the rest.

## Way 2 · Your regex: the same idea in ONE line
Way 1 needed a loop and an `if`. A regular expression says the same thing in one line:
**"a line that is only a year, then on the next line: word 1 (the make), a space, word 2 (the model)."**

You build it in **4 small steps**. Each step is a real, working pattern. The next cell runs every step on **file 03**
(the Elantra) and prints exactly what it matched, so you can see each piece do its job.

<table>
<tr><th>Step</th><th>Pattern so far</th><th>What the new piece means</th></tr>
<tr><td>1</td><td><code>^(?:19|20)\d{2}$</code></td><td>a line that is <b>only a year</b> starting with 19 or 20. <code>^</code> = start of a line, <code>$</code> = end of a line, <code>\d{2}</code> = two digits</td></tr>
<tr><td>2</td><td><code>^(?:19|20)\d{2}\n</code></td><td>swap <code>$</code> for <code>\n</code> (the line break): keep going onto the next line</td></tr>
<tr><td>3</td><td><code>^(?:19|20)\d{2}\n([A-Za-z-]+)</code></td><td><b>capture</b> the first word, letters and hyphens (<code>mercedes-benz</code>). The parentheses are what <code>.group(1)</code> returns</td></tr>
<tr><td>4</td><td><i>your turn:</i> add <code>\s+([A-Za-z0-9-]+)</code></td><td>a space (<code>\s+</code>), then <b>capture</b> the second word. Digits allowed, for models like <code>tl</code>, <code>x3</code>, <code>300-l</code></td></tr>
</table>

`re.MULTILINE` (at the end of the line) is what makes `^` and `$` mean "start/end of **each line**" instead of the whole ad.

## Way 3 · Stretch: fix the typos, flag the rest for review
Neither Way 1 nor Way 2 can know that `hyunda` means `hyundai`. A list of real makes, a few nicknames, and Python's
`difflib` (which finds the closest spelling) can. And when a "make" matches **nothing** on the list (like `contact`),
the honest answer is not a guess: it is **`NEEDS_REVIEW`**, so a person looks at it. That is how real data teams
handle the long tail.

````python
import difflib
from collections import Counter

KNOWN_MAKES = ["acura", "audi", "bmw", "buick", "cadillac", "chevrolet", "chrysler", "dodge", "fiat", "ford", "gmc",
               "honda", "hyundai", "infiniti", "jaguar", "jeep", "kia", "land", "lexus", "lincoln", "mazda",
               "mercedes-benz", "mercury", "mini", "mitsubishi", "nissan", "pontiac", "porsche", "ram", "saab",
               "saturn", "scion", "subaru", "tesla", "toyota", "volkswagen", "volvo"]
NICKNAMES = {"chevy": "chevrolet", "vw": "volkswagen", "mercedes": "mercedes-benz"}


def clean_make(make):
    make = str(make).strip().lower()
    if make in NICKNAMES:
        make = NICKNAMES[make]
    closest = difflib.get_close_matches(make, KNOWN_MAKES, n=1, cutoff=0.75)
    if len(closest) > 0:
        return closest[0]
    return "NEEDS_REVIEW"


def way3_make_model(text):
    make, model = way1_make_model(text)
    return clean_make(make), model


for i in range(len(graded)):
    text = read_ad_text(graded.iloc[i]["source_txt"])
    print("file", str(i + 1).zfill(2), "| extractor said:", graded.iloc[i]["make"], graded.iloc[i]["model"], "| way 3:", way3_make_model(text))
````

### How `difflib.get_close_matches` decides
```python
difflib.get_close_matches(word, possibilities, n=1, cutoff=0.75)
```
- **`word`**: the messy make you have, like `"hyunda"`
- **`possibilities`**: the clean list to pick from (`KNOWN_MAKES`)
- **`n`**: the most matches to return (`n=1` = only the single best one)
- **`cutoff`**: how similar a candidate must be, from `0` (anything goes) to `1` (identical). If nothing reaches the
  cutoff, you get an empty list `[]`, and that is when `clean_make` says `NEEDS_REVIEW`.

**Similarity** is a score from 0 to 1: roughly, 2 × (letters the two words share, in order) ÷ (letters in both words).
`hyunda` vs `hyundai`: they share 6 letters out of 6 + 7 = 13, so 2 × 6 ÷ 13 = **0.92**.

The next cell shows the top 3 scores for a few messy words, and what `get_close_matches` picks.

````python
CUTOFF = 0.75     # try 0.70 and watch what happens to "contact"

for word in ["hyunda", "toyta", "for", "chevy", "contact"]:
    scores = []
    for make in KNOWN_MAKES:
        similarity = difflib.SequenceMatcher(None, word, make).ratio()
        scores.append((round(similarity, 2), make))
    scores.sort(reverse=True)
    best = difflib.get_close_matches(word, KNOWN_MAKES, n=1, cutoff=CUTOFF)
    print(word.ljust(8), "| top 3:", scores[:3], "| picks:", best)
````

**Read the output:**
- `hyunda` → `hyundai` (0.92), `toyta` → `toyota` (0.91), `for` → `ford` (0.86): typos fixed.
- `chevy` scores only **0.57** against `chevrolet`, too different to pass. That is why `NICKNAMES` exists.
- `contact` scores **0.71** against `pontiac`, just under the cutoff, so it is flagged. **Set `CUTOFF = 0.70` and re-run:
  "Contact Information" becomes a Pontiac.**

Picking a cutoff is a trade-off: too high and real typos slip through; too low and the code invents cars. That is a
judgment call a data scientist makes, and should write down.

- **`reason`**: why: an **unknown make** (like `contact`), an **odd price** (under $500 or over $150,000), or a
  **placeholder mileage** (sellers who do not know type `99999`)
